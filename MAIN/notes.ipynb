
In Game-screen:

Asteroids
- Movement
  - After splitting -> new asteroids get bit of extra force and random direction (away from eachother)
- Sprite
- Split function
  - Big one -> 2 medium ones
  - Medium one -> two small ones
  - When limit reached -> don't split, instead spawn only one of lower level
- Collision with ship
  - Substract from total lives
- Collision with bullets
  - Add to score
- Limit: 26 asteroids on screen at a time

#### Asteroids life cycle
1. randomized spawn location outside of screen
2. randomized rotation
3. randomized speed
4. randomized direction (onto the screen)
5. if destroyed then spawn 2 of smaller category unless smallest
6. these smaller asteroids have randomized : rotation & speed & direction

Ship
- A/D for turning Left/Right respectfully, W for thrust (move in direction), Space for shooting
- Sprite
- Bullets spawm with Space
- Collision with asteroids

Bullets
- Movement
- Sprite
- Collision with asteroids

Lives
- Start 3 - 5
- Every 10.000 points extra one
- Cap amount of lives at a number (decide on it)
