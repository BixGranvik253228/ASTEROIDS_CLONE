```python
import arcade

class MyWindow(arcade.Window):
    def __init__(self, width, height, title):
        super().__init__(width, height, title)
        self.set_location(200, 0)
    
    def on_update(self, delta_time):
        print(delta_time)
        

MyWindow(1000, 900, 'Test Window')
arcade.run()
```
